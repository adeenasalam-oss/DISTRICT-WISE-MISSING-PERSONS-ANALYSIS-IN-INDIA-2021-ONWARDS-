# **DISTRICT-WISE MISSING PERSONS ANALYSIS IN INDIA (2021 ONWARDS)**

## **INTRODUCTION**
Missing persons cases remain a significant social and public safety concern in India, affecting individuals and families across different states and districts. Understanding the geographical distribution, age groups, and gender-wise trends of missing persons is essential for improving law enforcement strategies, public awareness, and policy formulation.

This project analyzes the **District-wise Missing Persons (2021 Onwards)** dataset to identify patterns in missing persons cases across India. Using Python-based data analysis and visualization techniques, the study explores state-wise and district-wise trends, compares gender and age-group distributions, and highlights regions with higher reported cases. The project demonstrates how data analytics can support evidence-based decision-making in public administration and crime analysis.

## **DATASET DESCRIPTION**
The dataset contains district-level records of missing persons reported across various states and union territories in India from **2021 onwards**. Each record represents the number of missing persons categorized by **gender** and **age group**, along with administrative details such as state, district, and registration circles.

The dataset enables detailed analysis of missing persons by location and demographic characteristics, making it valuable for identifying trends, regional variations, and vulnerable population groups.

## **DATASET SIGNIFICANCE**
This dataset is important because it provides detailed information on missing persons cases across India at the district level. It supports the identification of demographic trends based on age and gender while highlighting regional variations in reported cases.
The analysis can assist:
- Law enforcement agencies in identifying high-risk regions.
- Government authorities in policy planning and resource allocation.
- Researchers in studying crime and public safety trends.
- Social organizations in designing awareness and prevention initiatives.

* ## *CREATING DERIVED COLUMNS*

#### **1. TOTAL MISSING PERSONS**

```python
df['total_missing_persons'] = (
    df['male_children'] +
    df['male_18_years_and_above'] +
    df['female_children'] +
    df['female_18_years_and_above'] +
    df['transgender_children'] +
    df['transgender_18_years_and_above'])
df[['state_name', 'district_name', 'total_missing_persons']].head()
```

#### **2. TOTAL MALE, FEMALE AND TRANSGENDER CASES**

```python
df['total_male_missing'] = (
    df['male_children'] +
    df['male_18_years_and_above'])

df['total_female_missing'] = (
    df['female_children'] +
    df['female_18_years_and_above'])

df['total_transgender_missing'] = (
    df['transgender_children'] +
    df['transgender_18_years_and_above'])
```

#### **4. TOTAL CHILDREN AND ADULTS**

```python
df['total_children_missing'] = (
    df['male_children'] +
    df['female_children'] +
    df['transgender_children'])

df['total_adult_missing'] = (
    df['male_18_years_and_above'] +
    df['female_18_years_and_above'] +
    df['transgender_18_years_and_above'])
```

### ***<U>OBSERVATION</U>***
***The original dataset contains separate male, female, and transgender age-group counts. We can create meaningful analytical indicators from these variables.***

#### **5. AGE GROUP TOTALS**

```python
df['total_below_12'] = (
    df['male_below_12_years'] +
    df['female_below_12_years'] +
    df['transgender_below_12_years'])

df['total_12_to_16'] = (
    df['male_12_years_to_16_years'] +
    df['female_12_years_to_16_years'] +
    df['transgender_12_years_to_16_years'])

df['total_16_to_18'] = (
    df['male_16_years_to_18_years'] +
    df['female_16_years_to_18_years'] +
    df['transgender_16_years_to_18_years'])
```

### ***<U>OBSERVATION</U>***
***Because the dataset provides detailed age categories, we can also derive broader age-group indicators.***

# **VISUALIZATIONS**

```python
sns.set_theme(style="whitegrid")

black_red_palette = [
    '#000000',
    '#8B0000',
    '#B22222',
    '#CD5C5C',
    '#E6E6E6'
]
```

### **Visualization 1 — Year-wise Total Missing Persons**
#### *Purpose : To examine the change in total reported missing persons across the available years.*

```python
year_total = (
    df.groupby('year')['total_missing_persons']
      .sum()
      .reset_index()
)

plt.figure(figsize=(8,5))

sns.lineplot(
    data=year_total,
    x='year',
    y='total_missing_persons',
    marker='o',
    linewidth=2.5,
    color='#8B0000'
)

plt.title(
    'Year-wise Total Missing Persons',
    fontsize=14,
    fontweight='bold'
)

plt.xlabel('Year')
plt.ylabel('Total Missing Persons')

plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

### ***<U>OBSERVATION</U>***
***The line chart shows an increase in the total number of reported missing persons from 2021 to 2022. This indicates an upward movement in reported cases during the period covered by the dataset.***

### **Visualization 2 — Top 10 States by Total Missing Persons**
#### *Purpose : To identify the states reporting the highest aggregate number of missing persons.*

```python
top_10_states = (
    df.groupby('state_name')['total_missing_persons']
      .sum()
      .sort_values(ascending=False)
      .head(10)
)

plt.figure(figsize=(10,6))

sns.barplot(
    x=top_10_states.values,
    y=top_10_states.index,
    hue=top_10_states.index,
    palette=[
        '#000000',
        '#1A1A1A',
        '#333333',
        '#4D4D4D',
        '#660000',
        '#800000',
        '#8B0000',
        '#A52A2A',
        '#B22222',
        '#CD5C5C'
    ],
    legend=False
)

plt.title(
    'Top 10 States by Total Missing Persons',
    fontsize=14,
    fontweight='bold'
)

plt.xlabel('Total Missing Persons')
plt.ylabel('State')

plt.tight_layout()
plt.show()
```

### ***<U>OBSERVATION</U>***
***The visualization reveals considerable variation across states. Maharashtra records the highest aggregate number of missing persons, followed by other states with comparatively high reported totals. This indicates that missing-person cases are not evenly distributed geographically.***

### **Visualization 3 — Gender-wise Distribution of Missing Persons**

### ***<U>OBSERVATION</U>***

### **Visualization 6 — Distribution of Missing Persons by Gender**
#### *Purpose : To compare the spread and variation of missing-person counts across gender groups.*

```python
gender_df = df[
    [
        'total_male_missing',
        'total_female_missing',
        'total_transgender_missing'
    ]
].melt(
    var_name='Gender',
    value_name='Missing Persons'
)

gender_df['Gender'] = gender_df['Gender'].replace({
    'total_male_missing': 'Male',
    'total_female_missing': 'Female',
    'total_transgender_missing': 'Transgender'
})

plt.figure(figsize=(10,6))

sns.boxplot(
    data=gender_df,
    x='Gender',
    y='Missing Persons',
    hue='Gender',
    palette=[
        '#000000',
        '#8B0000',
        '#CD5C5C'
    ],
    legend=False
)

plt.title(
    'Distribution of Missing Persons by Gender',
    fontsize=14,
    fontweight='bold'
)

plt.xlabel('Gender')
plt.ylabel('Missing Persons')

plt.tight_layout()
plt.show()
```

### ***<U>OBSERVATION</U>***
***The box plot highlights substantial differences in the distribution of missing-person counts across gender groups. Female cases show a higher overall level and greater variation, while transgender cases remain comparatively low.***

#### *Purpose : To examine the relationship between male and female missing-person counts across district-level records.*

```python
plt.figure(figsize=(9,6))

sns.scatterplot(
    data=df,
    x='total_male_missing',
    y='total_female_missing',
    color='#8B0000',
    edgecolor='#000000',
    alpha=0.65
)

plt.title(
    'Relationship between Male and Female Missing Persons',
    fontsize=14,
    fontweight='bold'
)

plt.xlabel('Male Missing Persons')
plt.ylabel('Female Missing Persons')

plt.tight_layout()
plt.show()
```

### **Visualization 8 — Correlation Heatmap**
#### *Purpose : To examine relationships among the major missing-person indicators.*

```python
corr_cols = [
    'total_missing_persons',
    'total_male_missing',
    'total_female_missing',
    'total_transgender_missing',
    'total_children_missing',
    'total_adult_missing'
]

plt.figure(figsize=(10,7))

sns.heatmap(
    df[corr_cols].corr(),
    annot=True,
    cmap='Reds',
    fmt='.2f',
    linewidths=0.5,
    linecolor='white'
)

plt.title(
    'Correlation Matrix of Missing Person Indicators',
    fontsize=14,
    fontweight='bold'
)

plt.tight_layout()
plt.show()
```

### ***<U>OBSERVATION</U>***
***The heatmap shows strong positive relationships between total missing persons and the major gender- and age-based components. The strongest associations are observed with adult and female missing-person counts, while transgender cases show a comparatively weak relationship due to their much smaller magnitude.***

*Why Reds here?*

*The heatmap naturally needs multiple intensity levels. Using Reds gives us a white → light red → dark red scale, while the rest of the notebook uses black and red. This keeps the visualization consistent without making the correlation values difficult to read.*

### **Visualization 9 — Top 10 Districts by Total Missing Persons**

#### *Purpose : To identify districts with particularly high reported missing-person totals.*

```python
top_10_districts = (
    df.groupby(
        ['state_name', 'district_name']
    )['total_missing_persons']
    .sum()
    .sort_values(ascending=False)
    .head(10)
    .reset_index()
)

top_10_districts['District'] = (
    top_10_districts['district_name']
    + ', '
    + top_10_districts['state_name']
)

plt.figure(figsize=(11,6))

sns.barplot(
    data=top_10_districts,
    x='total_missing_persons',
    y='District',
    hue='District',
    palette=[
        '#000000',
        '#1A1A1A',
        '#333333',
        '#4D4D4D',
        '#660000',
        '#800000',
        '#8B0000',
        '#A52A2A',
        '#B22222',
        '#CD5C5C'
    ],
    legend=False
)

plt.title(
    'Top 10 Districts by Total Missing Persons',
    fontsize=14,
    fontweight='bold'
)

plt.xlabel('Total Missing Persons')
plt.ylabel('District')

plt.tight_layout()
plt.show()
```

### ***<U>OBSERVATION</U>***
***The district-level visualization demonstrates considerable geographical variation. A relatively small number of districts account for very high reported totals, highlighting areas that warrant closer examination.***

### **Visualization 10 — Year-wise Gender Comparison**

#### *Purpose : To compare male, female, and transgender missing-person cases across the available years.*

```python
year_gender = (
    df.groupby('year')[
        [
            'total_male_missing',
            'total_female_missing',
            'total_transgender_missing'
        ]
    ]
    .sum()
    .reset_index()
)

year_gender_melted = year_gender.melt(
    id_vars='year',
    var_name='Gender',
    value_name='Missing Persons'
)

year_gender_melted['Gender'] = (
    year_gender_melted['Gender'].replace({
        'total_male_missing': 'Male',
        'total_female_missing': 'Female',
        'total_transgender_missing': 'Transgender'
    })
)

plt.figure(figsize=(10,6))

sns.barplot(
    data=year_gender_melted,
    x='year',
    y='Missing Persons',
    hue='Gender',
    palette=[
        '#000000',
        '#8B0000',
        '#CD5C5C'
    ]
)

plt.title(
    'Year-wise Gender Comparison of Missing Persons',
    fontsize=14,
    fontweight='bold'
)

plt.xlabel('Year')
plt.ylabel('Missing Persons')

plt.legend(
    title='Gender',
    loc='upper left'
)

plt.tight_layout()
plt.show()
```

### ***<U>OBSERVATION</U>***
***Female cases remain substantially higher than male cases across the years covered by the dataset. Transgender cases remain comparatively small. The grouped comparison clearly illustrates the gender-wise composition and its change across the reporting years.***
